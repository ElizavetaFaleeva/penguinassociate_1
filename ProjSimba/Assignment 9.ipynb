```python
text = ("Before departure, all students leaving for exchange studies should fill in "
        "a specialized application form. An application should be submitted in 1 copy "
        "and not later than 2 weeks prior to departure for the exchange semester, "
        "provided that the students do not have any academic failure. The paper should "
        "be submitted to the International Academic Mobility Office (Volkhovsky per., 3, "
        "office 207, Tel.: +7 (812) 323 8447). A copy of the invitation letter from a "
        "host school - partner of GSOM SPbU has to be attached to the application.")

open_paren = text.find("(")
print("First '(' at index:", open_paren)

tel_pos = text.find(", Tel.", open_paren)
print("', Tel.'  at index:", tel_pos)

address = text[open_paren + 1 : tel_pos]

print("Office address:", address)
```

```python


print("=" * 60)
print("LISTS")
print("=" * 60)

fruits = ["apple", "banana", "cherry", "banana"]
print("Original :", fruits)
print("Length   :", len(fruits))

fruits.append("date")                
fruits.insert(1, "apricot")          
fruits.remove("banana")               
last = fruits.pop()                  
print("After modifications:", fruits)
print("Popped item :", last)
print("Index of 'cherry':", fruits.index("cherry"))
print("Count of 'banana':", fruits.count("banana"))
fruits.sort()
print("Sorted   :", fruits)
fruits.reverse()
print("Reversed :", fruits)


print("Slice [1:4] :", fruits[1:4])


print("\n" + "=" * 60)
print("TUPLES")
print("=" * 60)

point = (10, 20, 30)
print("Tuple       :", point)
print("First elem  :", point[0])
print("Last elem   :", point[-1])
print("Slice [1:]  :", point[1:])
print("Length      :", len(point))


x, y, z = point
print(f"Unpacked: x={x}, y={y}, z={z}")


try:
    point[0] = 99
except TypeError as e:
    print("Tuples are immutable:", e)


print("\n" + "=" * 60)
print("DICTIONARIES")
print("=" * 60)

student = {"name": "Alex", "age": 20, "gpa": 4.5}
print("Dict       :", student)

student["major"] = "Management"      
student["age"] = 21                  
print("After edits:", student)
print("Keys       :", list(student.keys()))
print("Values     :", list(student.values()))
print("Items      :", list(student.items()))
print("Get 'name' :", student.get("name"))
print("Get missing:", student.get("height", "not set"))
removed = student.pop("gpa")
print("Popped gpa :", removed)
print("Final dict :", student)


print("\n" + "=" * 60)
print("SETS")
print("=" * 60)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("A           :", a)
print("B           :", b)
print("Union       :", a | b)
print("Intersection:", a & b)
print("Difference  :", a - b)
print("Sym. diff   :", a ^ b)

a.add(7)
a.discard(1)
print("A after add/discard:", a)


dup = {1, 1, 2, 2, 3}
print("Duplicates removed:", dup)


print("\n" + "=" * 60)
print("NESTED COLLECTIONS")
print("=" * 60)


matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Matrix          :", matrix)
print("matrix[1][2]    :", matrix[1][2])     # 6
matrix[0][0] = 99
print("After edit      :", matrix)


team = [
    {"name": "Alex",   "role": "dev"},
    {"name": "Maria",  "role": "pm"},
    {"name": "Ivan",   "role": "qa"},
]
print("\nTeam members:")
for member in team:
    print(f"  {member['name']:>6}  ->  {member['role']}")


courses = {
    "math":    [5, 4, 5],
    "history": [3, 4],
    "python":  [5, 5, 5],
}
print("\nCourses and grades:")
for course, grades in courses.items():
    avg = sum(grades) / len(grades)
    print(f"  {course:>8}: {grades}  avg={avg:.2f}")


university = {
    "GSOM": {
        "dean": "Olga Dergunova",
        "students": 1200,
        "programs": ["Bachelor", "Master", "MBA"],
    },
    "Math faculty": {
        "dean": "Someone Else",
        "students": 800,
        "programs": ["Bachelor", "Master"],
    },
}
print("\nUniversity structure:")
for faculty, info in university.items():
    print(f"  {faculty}: {info['students']} students, dean={info['dean']}")


pairs = {(1, 2), (3, 4), (1, 2)}   
print("\nSet of tuples (duplicate dropped):", pairs)


mixed = (1, "two", [3, 4])
print("Original tuple:", mixed)

mixed[2].append(5)
print("After list.append:", mixed)

try:
    mixed[0] = 999
except TypeError as e:
    print("Cannot change tuple element:", e)
```
